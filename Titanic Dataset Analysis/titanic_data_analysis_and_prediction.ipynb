## About the Data:

The sinking of the Titanic is one of the most infamous shipwrecks in history.

On April 15, 1912, during her maiden voyage, the widely considered “unsinkable” RMS Titanic sank after colliding with an iceberg. Unfortunately, there weren’t enough lifeboats for everyone onboard, resulting in the death of 1502 out of 2224 passengers and crew.

While there was some element of luck involved in surviving, it seems some groups of people were more likely to survive than others.

In this challenge, we ask you to build a predictive model that answers the question: “what sorts of people were more likely to survive?” using passenger data (ie name, age, gender, socio-economic class, etc).

### Deal with missing values

#### "Cabin"

As the percentage of missing values in the "Cabin column" is very high:
- There are 77% missing values.
- Imputations make it impractical
- we can simply drop the missing values.

```python
df_train.drop('Cabin', axis=1, inplace=True)
df_test.drop('Cabin', axis=1, inplace=True)
```

#### "Age"

```python
from sklearn.impute import KNNImputer

# Select numeric columns for KNN imputation (excluding target 'Survived' if present)
knn_features = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

# KNN imputation for df_train
imputer = KNNImputer(n_neighbors=5)
df_train_impute = df_train.copy()
df_train_impute[knn_features] = imputer.fit_transform(df_train[knn_features])

# KNN imputation for df_test
df_test_impute = df_test.copy()
df_test_impute[knn_features] = imputer.transform(df_test[knn_features])


# Replace original Age columns with imputed values
df_train['Age'] = df_train_impute['Age']
df_test['Age'] = df_test_impute['Age']
```

#### "Embarked"

```python
# Fill missing values in 'Embarked' column with mode
df_train['Embarked'].fillna(df_train['Embarked'].mode()[0], inplace=True)
# Fill missing values in 'Fare' column with mean
df_test['Fare'].fillna(df_test['Fare'].mean(), inplace=True)
```

### Fare:

    - Positively correlated with Survived (+0.26): Higher ticket fare → better survival rate.

    - Negatively correlated with Pclass (-0.55): Lower class → lower fare.

✅ Good predictor for socioeconomic status and survival.

## Feature Engineering

### Drop irrelevant features

```python
df_train.drop(columns=['PassengerId', 'Name', 'Ticket'], inplace=True)
df_test.drop(columns=['PassengerId', 'Name', 'Ticket'], inplace=True)
```

```python
df_train.info()
```

```python
df_test.info()
```

Converting categorical features to numerical features

```python
# Fit the encoder on the training data only 
# Transform both training and test data

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

le.fit(df_train['Sex'])
df_train['Sex'] = le.transform(df_train['Sex'])
df_test['Sex'] = le.transform(df_test['Sex'])

le.fit(df_train['Embarked'])
df_train['Embarked'] = le.transform(df_train['Embarked'])
df_test['Embarked'] = le.transform(df_test['Embarked'])
```

**Feature Scaling**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

Numeric_features = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

# Fit the scaler on the training data only
scaler.fit(df_train[Numeric_features])
df_train[Numeric_features] = scaler.transform(df_train[Numeric_features])
df_test[Numeric_features] = scaler.transform(df_test[Numeric_features])
```

**Separate the features and the target variable**

```python
x_train = df_train.drop('Survived', axis=1)
y_train = df_train['Survived']
x_test = df_test.copy()
```

## Model Implementation

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 1. Initialize the model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# 2. Train the model on training data
rf_model.fit(x_train, y_train)

# 3. Predict on the training set (to check performance)
y_train_pred = rf_model.predict(x_train)

# 4. Print evaluation on training data
print("Training Accuracy:", accuracy_score(y_train, y_train_pred))
print("Confusion Matrix:\n", confusion_matrix(y_train, y_train_pred))
print("Classification Report:\n", classification_report(y_train, y_train_pred))

# 5. Predict on test data (actual prediction)
y_test_pred = rf_model.predict(x_test)


```

```python
#Save predictions for submission

# Assuming df_test has PassengerId column
submission = pd.DataFrame({
    'PassengerId': df_test_original['PassengerId'],
    'Survived': y_test_pred
})

submission.to_csv('random_forest_submission.csv', index=False)
```
