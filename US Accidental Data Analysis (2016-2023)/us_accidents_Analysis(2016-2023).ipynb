There are many missing values in the dataset. We will deal with them using the following steps:
As there are five features that have missing values in massive amounts as compare to the dataset size

let's fill out these missing value using the following steps:

A) End_Latitude and End_Longitude

```python
# Impute End_Lat using the mean End_Lat grouped by Start_Lat
data['End_Lat'] = data['End_Lat'].fillna(data.groupby('Start_Lat')['End_Lat'].transform('mean'))

# Impute End_Lng using the mean End_Lng grouped by Start_Lng
data['End_Lng'] = data['End_Lng'].fillna(data.groupby('Start_Lng')['End_Lng'].transform('mean'))

# Check the number of missing values
data.isnull().sum()
```

Still there are missing values in the End_Lat and End_Lng columns. We will impute them using feature "Distance(mi)".

```python
# Use distance-based grouping to fill missing End_Lat and End_Lng values with the mean of their respective Distance(mi) groups
data['End_Lat'] = data.groupby('Distance(mi)')['End_Lat'].transform(lambda x: x.fillna(x.mean()))
data['End_Lng'] = data.groupby('Distance(mi)')['End_Lng'].transform(lambda x: x.fillna(x.mean()))

# Check the number of missing values
data.isnull().sum()
```

Now we have 5499 missing values in End_Lat and 7928 missing values in End_Lng(Too small as compare to dataset).

Simply fill it by using mean of End_Lat and End_Lng.

```python
# Fill the remaining missing values with the mean of the column
data['End_Lat'] = data['End_Lat'].fillna(data['End_Lat'].mean())
data['End_Lng'] = data['End_Lng'].fillna(data['End_Lng'].mean())

# Check the number of missing values
data.isnull().sum()
```

There is no missing value left in End_Lat and End_Lng.

Let's move to the next feature:

B) Wind_Chill(F)

Wind chill is a feels-like temperature:

1) Actual air temperature (→ Temperature(F))


```python
data['Wind_Chill(F)'] = data.groupby(['Temperature(F)'])['Wind_Chill(F)'].transform(lambda x: x.fillna(x.mean()))

# Check the number of missing values
data.isnull().sum()
```

Still we are left with more than 70k missing values in Wind_Chill(F)

Let's fill it by using mean per Weather_Condition

```python
# Fill remaining missing values with group-wise mean by Weather_Condition
data['Wind_Chill(F)'] = data['Wind_Chill(F)'].fillna(
    data.groupby('Weather_Condition')['Wind_Chill(F)'].transform('mean')
)
```

C) Wind_Speed(mph)

```python
# Fill missing Wind_Speed using mean per Weather_Condition
data['Wind_Speed(mph)'] = data['Wind_Speed(mph)'].fillna(
    data.groupby('Weather_Condition')['Wind_Speed(mph)'].transform('mean')
)
```

D) Precipitation(in)

```python
# Fill missing Precipitation(in) using mean per Weather_Condition

data['Precipitation(in)'] = data.groupby('Weather_Condition')['Precipitation(in)'].transform(lambda x: x.fillna(x.mean()))

```
